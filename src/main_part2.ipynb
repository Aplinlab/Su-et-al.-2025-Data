# SIMPLE SPIKERATE (per unit)

```python
# User defined variables
save_figures = True


### User does not need to modify below this line ###

ssr_df = hff_analysis.calculate_ssr(spikes_df)
for unit_id in ssr_df['Unit ID'].unique():
    unit_df = ssr_df.loc[ssr_df['Unit ID'] == unit_id]
    unit_type = hff_analysis.unique(unit_df['Unit Type'])
    hff_analysis.plot_combined_ssr(
        unit_df,
        f"{unit_id.upper().replace('_', '-')} ({unit_type})",
        save_figures,
        unit_id
    )
```

# SIMPLE SPIKERATE (all units)

# RESPONSE RATE vs TIME (per unit)
This is a temporary block of code to generate these plots for the conditioning phase only.

```python
import matplotlib.pyplot as plt
# User defined variables
conditioning_df = spikes_df.loc[spikes_df['Phase'] == 'conditioning']
for unit_id in conditioning_df['Unit ID'].unique():
    unit_df = conditioning_df.loc[conditioning_df['Unit ID'] == unit_id]
    for test in unit_df['Test'].unique():
        test_df = unit_df.loc[unit_df['Test'] == test]
        for stim_type in test_df['Test Stimulus'].unique():
            plt.figure()
            hff_analysis.spikequantify.plot_single_rvt(
                test_df,
                unit_id,
                test,
                stim_type,
                'conditioning'
            )
```
