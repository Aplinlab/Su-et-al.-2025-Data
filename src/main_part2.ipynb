# STEP 0: INITIALISATION

Run this code block once per session, to initialise and import modules
and get a list of JSON files created in `main_part1.ipynb`.

```python
import hff_analysis
from os import walk
import pathlib

json_spike_filenames = [
    f for f in next(walk(hff_analysis.SPIKES_JSON_FOLDER), (None, None, []))[2]
    if f.endswith(".json")
]
for filename in json_spike_filenames:
    print(f"{filename}")
```

# STEP 1: SIMPLE PLOT (FREQUENCY SWEEPS)

```python
# User defined variables
save_figures = True


### User does not need to modify below this line ###

simple_spikerate_df = hff_analysis.simple_spikerate_df(json_spike_filenames)
simple_plot_df = hff_analysis.simple_spikerate_plotdf(simple_spikerate_df)

try:
    hff_analysis.plot_simple_spikerate(
        simple_plot_df,
        'Conditioning',
        save_figures
    )
except FileNotFoundError:
    pathlib.Path(hff_analysis.PLOTS_FOLDER).mkdir(
        parents=True,
        exist_ok=True
    )
    hff_analysis.plot_simple_spikerate(
        simple_plot_df,
        'Conditioning',
        save_figures
    )

hff_analysis.plot_simple_spikerate(
    simple_plot_df,
    'Interleaved',
    save_figures
)

hff_analysis.plot_simple_spikerate(
    simple_plot_df,
    'Recovery',
    save_figures
)
```
