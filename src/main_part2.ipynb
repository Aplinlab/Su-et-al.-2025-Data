```python
import matplotlib.pyplot as plt
import hff_analysis

# User defined variables
load_df_json = True


### User does not need to modify below this line ###

spikes_df = hff_analysis.spikes_table(load_df_json)
spikes_df
```

# SIMPLE PLOT (FREQUENCY SWEEPS INDIVIDUAL)

```python
# User defined variables
save_figures = True


### User does not need to modify below this line ###

freq_df = spikes_df.loc[spikes_df['Test'] == 'Frequency']
for unit_id in freq_df['Unit ID'].unique():
    ssr_df = hff_analysis.simple_spikerate_df(
        freq_df.loc[freq_df['Unit ID'] == unit_id]
    )
    unit_type, *unit_type_empty = ssr_df['Unit Type'].unique()
    assert len(unit_type_empty) == 0

    ssr_plot_df = hff_analysis.simple_spikerate_plotdf(ssr_df)

    hff_analysis.plot_simple_spikerate(
        ssr_plot_df,
        'Conditioning',
        f'{unit_id} ({unit_type}) - Conditioning',
        save_figures,
        unit_id + "_Conditioning"
    )

    hff_analysis.plot_simple_spikerate(
        ssr_plot_df,
        'Interleaved',
        f'{unit_id} ({unit_type}) - Interleaved',
        save_figures,
        unit_id + "_Interleaved"
    )

    hff_analysis.plot_simple_spikerate(
        ssr_plot_df,
        'Recovery',
        f'{unit_id} ({unit_type}) - Recovery',
        save_figures,
        unit_id + "_Recovery"
    )
```

# SIMPLE PLOT (FREQUENCY SWEEPS COMBINED)

```python
# User defined variables
save_figures = True


### User does not need to modify below this line ###

freq_df = spikes_df.loc[spikes_df['Test'] == 'Frequency']
ssr_df = hff_analysis.simple_spikerate_df(freq_df)
ssr_plot_df = hff_analysis.simple_spikerate_plotdf(ssr_df)

hff_analysis.plot_simple_spikerate(
    ssr_plot_df,
    'Conditioning',
    f'{unit_id} ({unit_type}) - Conditioning',
    save_figures,
    unit_id + "_Conditioning"
)

hff_analysis.plot_simple_spikerate(
    ssr_plot_df,
    'Interleaved',
    f'{unit_id} ({unit_type}) - Interleaved',
    save_figures,
    unit_id + "_Interleaved"
)

hff_analysis.plot_simple_spikerate(
    ssr_plot_df,
    'Recovery',
    f'{unit_id} ({unit_type}) - Recovery',
    save_figures,
    unit_id + "_Recovery"
)
```

# RASTER PLOT (FREQUENCY SWEEPS)

```python
# User defined variables
save_figures = True

### User does not need to modify below this line ###

freq_df = spikes_df.loc[spikes_df['Test'] == 'Frequency']
plt.figure(figsize=hff_analysis.FIGSIZE)
frequency_list = sorted(freq_df['Test Frequency'].unique())
for plot_id, frequency in enumerate(frequency_list):
    frequency_df = freq_df.loc[freq_df['Test Frequency'] == frequency]
    hff_analysis.plot_raster(
        frequency_df,
        plot_id,
        len(frequency_list),
        'Trial ID',
        save_figures
    )
```
